```python
NCAA_PATH = r"Data\27-4-2025_player.xlsx"
ncaa = pd.read_excel(NCAA_PATH)
```

```python
#create a column for merging the dbs
ncaa['UID_scout'] = ncaa['Player'].str.strip().str.lower().str.replace(' ', '') + ncaa['Team'].str.lower().str[0:3]
ncaa['team_for_merging'] = ncaa['Team'].str.strip().str.lower().str.replace(' ', '').str.replace('-', '')
```

## Scouting reports

Need the creation of a UID that is a combination between the players' name and the first three character of the team.

```python
SCOUTING_REPORT = r"Data\Scouting reports - Copie (2).xlsx" #Copie is the one without double name
scout = pd.read_excel(SCOUTING_REPORT, sheet_name="Scouting Report 2006-2025")
```

```python
#Smaller cases for the players' name
scout['UID_scout'] = scout['Name_for_merging'].str.lower() + scout['College'].str.lower().str[0:3]
#Merging on the correspondance of Player name, however we choose how outer to leave the not corresponding row empty
working_df = pd.merge(ncaa, scout, on=['UID_scout'], how='left') 
```

## Team data

```python
TEAM_PATH = r"Data\27-4-2025_team.xlsx"
team = pd.read_excel(TEAM_PATH)
```

```python
colonne = 'team_for_merging'  
uniques_df1 = set(ncaa[colonne].unique())
#uniques_df2 = set(team[colonne].unique())

#Does all the teams from df1 are in df2?
#diff_df1_df2 = uniques_df1 - uniques_df2  
#diff_df2_df1 = uniques_df2 - uniques_df1
```

```python
#Smaller cases for the players' name
team['team_for_merging'] = team['School'].str.strip(
                            ).str.lower().str.replace(' ', '').str.replace('ncaa', ''
                            ).str.replace('-', '').str.replace('&', '').str.replace('(', '').str.replace(')', ''
                            ).str.replace('.', '').str.replace("'", "")
#Merging on the correspondance of Player name, however we choose how outer to leave the not corresponding row empty
working_df = pd.merge(working_df, team, on=['team_for_merging', 'season'], 
                     suffixes=['', '_team'], how='left') 
```

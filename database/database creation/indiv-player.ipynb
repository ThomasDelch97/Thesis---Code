# Data merging and concatenation for team stats

```python
#Does the df have the same length
dic_of_col = {}
conds = [
    'school',
    'opponent'
]
for year in range(2000, 2026):
    for cond in conds:
        try:
            working_file = rf"\Team NCAA {year} {cond}.csv"
            PATH = HOME + TEAM + working_file
            df = pd.read_csv(PATH)
            dic_of_col[(year, cond)] = df.shape
        except:
            continue
dic_of_col
```

```python
container = 0
for i, j in dic_of_col.items():
    if container > 0:
        container = container + j[0]  
    else: 
        container = j[0]
container
```

```python
dic_of_df = {}
for year in range(2000, 2026):
    for i, cond in enumerate(conds):
        working_file = rf"\Team NCAA {year} {cond}.csv"
        PATH = HOME + TEAM + working_file
        try:
            if i == 0:
                df = pd.read_csv(PATH)
                df['season'] = year
            else:
                df1 = pd.read_csv(PATH)
                df1['season'] = year
                df = pd.merge(df, df1, on=['School', 'season'], suffixes=['', '_opp'])
            dic_of_df[year] = df.copy()
        except:
            continue

for year, df in dic_of_df.items():
    if year == min(dic_of_df.keys()):
        working_df = df.copy()
    else:
        working_df = pd.concat([working_df, df])
```

```python
working_df.to_excel("27-4-2025_team.xlsx")
```
